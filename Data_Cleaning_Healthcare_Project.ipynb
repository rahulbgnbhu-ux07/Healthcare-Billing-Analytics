```python
import pandas as pd
import zipfile

# Extract Healthcare dataset
with zipfile.ZipFile("healthcare_dataset.csv.zip", "r") as zip_ref:
    zip_ref.extractall()

# Extract Appointment dataset
with zipfile.ZipFile("KaggleV2-May-2016.csv.zip", "r") as zip_ref:
    zip_ref.extractall()

# Read the CSV files
healthcare = pd.read_csv("healthcare_dataset.csv")
appointments = pd.read_csv("KaggleV2-May-2016.csv")

print("Healthcare dataset:", healthcare.shape)
print("Appointment dataset:", appointments.shape)
```

```python
healthcare.head()
```

```python
appointments.head()
```

```python
print("Healthcare Dataset")
print(healthcare.head())
print("\nColumns:")
print(healthcare.columns.tolist())

print("\n" + "="*60)

print("Appointment Dataset")
print(appointments.head())
print("\nColumns:")
print(appointments.columns.tolist())
```

```python
print("Healthcare data types:")
print(healthcare.dtypes)

print("\nAppointment data types:")
print(appointments.dtypes)
```

```python
print("Healthcare Dataset")
print("Missing values:")
print(healthcare.isnull().sum())

print("\nDuplicate rows:")
print(healthcare.duplicated().sum())

print("\n" + "="*50)

print("Appointment Dataset")
print("Missing values:")
print(appointments.isnull().sum())

print("\nDuplicate rows:")
print(appointments.duplicated().sum())
```

```python
# Number of times each complete row appears
duplicate_counts = (
    healthcare[healthcare.duplicated(keep=False)]
    .value_counts()
)

print(duplicate_counts.head(20))
```

```python
duplicate_counts = (
    healthcare[healthcare.duplicated(keep=False)]
    .value_counts()
)

print("Number of duplicated unique records:", len(duplicate_counts))

print("\nHow many records appear 2, 3, 4... times?")
print(duplicate_counts.value_counts().sort_index())
```

```python
# Compare basic statistics before removing duplicates

print("Original dataset:")
print("Rows:", len(healthcare))
print("Total billing:", healthcare["Billing Amount"].sum())
print("Average billing:", healthcare["Billing Amount"].mean())

print("\nAfter removing exact duplicates:")

healthcare_temp = healthcare.drop_duplicates()

print("Rows:", len(healthcare_temp))
print("Total billing:", healthcare_temp["Billing Amount"].sum())
print("Average billing:", healthcare_temp["Billing Amount"].mean())
```

```python
healthcare_clean = healthcare.drop_duplicates().copy()
```

```python
print("AGE")
print("Minimum:", healthcare_clean["Age"].min())
print("Maximum:", healthcare_clean["Age"].max())

print("\nBILLING")
print("Minimum:", healthcare_clean["Billing Amount"].min())
print("Maximum:", healthcare_clean["Billing Amount"].max())

print("\nADMISSION TYPES")
print(healthcare_clean["Admission Type"].value_counts())

print("\nTEST RESULTS")
print(healthcare_clean["Test Results"].value_counts())

print("\nMEDICATIONS")
print(healthcare_clean["Medication"].value_counts())
```

```python
negative_billing = healthcare_clean[
    healthcare_clean["Billing Amount"] < 0
]

print("Negative billing records:", len(negative_billing))

print("\nNegative billing values:")
print(negative_billing["Billing Amount"].describe())

print("\nNegative billing records:")
display(negative_billing)
```

```python
print("Negative billing records:", len(negative_billing))

print("\nUnique negative billing values:",
      negative_billing["Billing Amount"].nunique())

print("\nNegative billing by admission type:")
print(negative_billing["Admission Type"].value_counts())
```

```python
healthcare_clean["Billing_Valid"] = healthcare_clean["Billing Amount"].where(
    healthcare_clean["Billing Amount"] >= 0
)
```

```python
appointments_clean = appointments.copy()

print("Appointment dataset copied successfully.")
print("Rows:", len(appointments_clean))
```

```python
print("Admission date examples:")
print(healthcare_clean["Date of Admission"].head())

print("\nDischarge date examples:")
print(healthcare_clean["Discharge Date"].head())

print("\nAppointment scheduled date examples:")
print(appointments_clean["ScheduledDay"].head())

print("\nAppointment date examples:")
print(appointments_clean["AppointmentDay"].head())
```

```python
# Convert Healthcare dates
healthcare_clean["Date of Admission"] = pd.to_datetime(
    healthcare_clean["Date of Admission"]
)

healthcare_clean["Discharge Date"] = pd.to_datetime(
    healthcare_clean["Discharge Date"]
)

# Convert Appointment dates
appointments_clean["ScheduledDay"] = pd.to_datetime(
    appointments_clean["ScheduledDay"],
    utc=True
)

appointments_clean["AppointmentDay"] = pd.to_datetime(
    appointments_clean["AppointmentDay"],
    utc=True
)

print("Dates converted successfully!")
```

```python
print(healthcare_clean[
    ["Date of Admission", "Discharge Date"]
].dtypes)

print("\nAppointment dates:")
print(appointments_clean[
    ["ScheduledDay", "AppointmentDay"]
].dtypes)
```

```python
healthcare_clean["Discharge Date"] = pd.to_datetime(
    healthcare_clean["Discharge Date"],
    format="mixed"
)

print(healthcare_clean["Discharge Date"].dtype)
```

```python
print(healthcare_clean[
    ["Date of Admission", "Discharge Date"]
].dtypes)

print("\nAppointment dates:")
print(appointments_clean[
    ["ScheduledDay", "AppointmentDay"]
].dtypes)
```

```python
invalid_dates = healthcare_clean[
    healthcare_clean["Discharge Date"] < healthcare_clean["Date of Admission"]
]

print("Records with discharge before admission:", len(invalid_dates))

display(
    invalid_dates[
        ["Date of Admission", "Discharge Date"]
    ].head(10)
)
```

```python
healthcare_clean["Length_of_Stay"] = (
    healthcare_clean["Discharge Date"]
    - healthcare_clean["Date of Admission"]
).dt.days

print(healthcare_clean["Length_of_Stay"].describe())
```

```python
healthcare_clean["Valid_Billing"] = healthcare_clean["Billing Amount"].where(
    healthcare_clean["Billing Amount"] >= 0
)
```

```python
healthcare_clean["Valid_Billing"] = healthcare_clean["Billing Amount"].where(
    healthcare_clean["Billing Amount"] >= 0
)

print("Total records:", len(healthcare_clean))
print("Valid billing records:", healthcare_clean["Valid_Billing"].notna().sum())
print("Invalid billing records:", healthcare_clean["Valid_Billing"].isna().sum())
```

```python
print("AGE")
print("Minimum:", appointments_clean["Age"].min())
print("Maximum:", appointments_clean["Age"].max())

print("\nNO-SHOW")
print(appointments_clean["No-show"].value_counts())

print("\nGENDER")
print(appointments_clean["Gender"].value_counts())

print("\nHEALTH CONDITIONS")
print("Hypertension:")
print(appointments_clean["Hipertension"].value_counts())

print("\nDiabetes:")
print(appointments_clean["Diabetes"].value_counts())

print("\nAlcoholism:")
print(appointments_clean["Alcoholism"].value_counts())

print("\nHandicap:")
print(appointments_clean["Handcap"].value_counts())

print("\nSMS Received:")
print(appointments_clean["SMS_received"].value_counts())
```

```python
# Check how many records have Age = -1
age_minus_one = appointments_clean[
    appointments_clean["Age"] == -1
]

print("Records with Age = -1:", len(age_minus_one))

display(age_minus_one)
```

```python
appointments_clean.loc[
    appointments_clean["Age"] == -1, "Age"
] = pd.NA
```

```python
print("Age = -1 remaining:",
      (appointments_clean["Age"] == -1).sum())

print("Missing age values:",
      appointments_clean["Age"].isna().sum())
```

```python
age_115 = appointments_clean[
    appointments_clean["Age"] == 115
]

print("Records with Age = 115:", len(age_115))
display(age_115)
```

```python
appointments_clean.loc[
    appointments_clean["Age"] == 115, "Age"
] = pd.NA
```

```python
print("Age = -1 remaining:",
      (appointments_clean["Age"] == -1).sum())

print("Age = 115 remaining:",
      (appointments_clean["Age"] == 115).sum())

print("Missing age values:",
      appointments_clean["Age"].isna().sum())
```

```python
appointments_clean["Waiting_Days"] = (
    appointments_clean["AppointmentDay"].dt.normalize()
    - appointments_clean["ScheduledDay"].dt.normalize()
).dt.days

print(appointments_clean["Waiting_Days"].describe())
```

```python
negative_waiting = appointments_clean[
    appointments_clean["Waiting_Days"] < 0
]

print("Negative waiting-time records:", len(negative_waiting))

display(
    negative_waiting[
        [
            "PatientId",
            "AppointmentID",
            "ScheduledDay",
            "AppointmentDay",
            "Waiting_Days",
            "No-show"
        ]
    ].head(20)
)
```

```python
appointments_clean.loc[
    appointments_clean["Waiting_Days"] < 0,
    "Waiting_Days"
] = pd.NA
```

```python
print("Negative waiting times remaining:",
      (appointments_clean["Waiting_Days"] < 0).sum())

print("Missing waiting times:",
      appointments_clean["Waiting_Days"].isna().sum())
```

```python
# Create final cleaned datasets

healthcare_final = healthcare_clean.copy()
appointments_final = appointments_clean.copy()

print("Healthcare final shape:", healthcare_final.shape)
print("Appointments final shape:", appointments_final.shape)
```

```python
healthcare_final.to_csv(
    "healthcare_clean.csv",
    index=False
)

appointments_final.to_csv(
    "appointments_clean.csv",
    index=False
)

print("Clean datasets saved successfully.")
```
